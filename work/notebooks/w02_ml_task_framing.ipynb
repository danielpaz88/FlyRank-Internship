## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

My lane is a **ranking** problem (specifically, a **scoring** problem whose output creates a ranking).

**Why not classification?**  
Classification would ask: "Will this district flip? (Yes/No)." That loses nuance — some districts have *more* potential to change than others. Chris (my target audience) doesn't just need a binary list; he needs priorities.

**Why not clustering?**  
Clustering would group districts into segments (e.g., "urban strongholds," "rural swing districts"). That's useful for discovery, but Chris (my target audience) needs a *sorted list* to allocate sequential resources (first to district #1, then #2, etc.).

**Why ranking/scoring?**  
Chris has limited resources (time, money, staff). He needs to know exactly *which district to invest in first*. A score for each district allows me to sort them from "highest predicted change" to "lowest." He then draws a cut based on his budget.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

The primary success metric is **Precision@K** (Top-K Precision).

**What is K?**  
K is the **number of districts Chris has the budget or capacity to actively target**. If Chris can only run an intensive campaign in 5 districts, K=5. If he has resources for 20, K=20.

**What does Precision@K measure?**  
It answers: *"Of the K districts my model recommends as the highest potential for change, how many of them actually are in the real top K districts based on actual ΔVoteShare?"*

**Formula:**

> Precision@K = (Number of correctly identified top-K districts) / K

**Why this metric over RMSE or R²?**  
Chris doesn't need the exact Δ value (e.g., "+6.7%"). He needs the *correct ranking* of priorities. A model with slightly off scores but perfect ranking (e.g., it predicts the exact top 5, even if scores are +8% vs +7%) is far more useful than a model with accurate scores but wrong order.

**What number means "good"?**  
For Salta, Argentina (23 departments), if K=5:
- Precision@5 = 1.0 (perfect: the top 5 predicted are the actual top 5) → Excellent.
- Precision@5 = 0.8 (4 predicted out of 5 correct) → Good.
- Precision@5 = 0.4 (2 predicted out of 5 correct) → Needs improvement.

The baseline (random guessing) would be K/Total = 5/23 ≈ 0.22. Anything above 0.5 is already beating random by a wide margin.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule might be: *"Focus campaign resources on the districts with the highest population growth."*

This rule is simple and intuitive, but it fails because:

1. **Population growth ≠ vote change.** A district can grow but be full of loyal opposition voters.
2. **Rules are linear and ignore interactions.** A rule cannot capture: *"High poverty + low education + high historical opposition vote = high change potential"* — but *only* if a new candidate runs with a strong social message.
3. **Rules don't adapt.** If new census data or polling comes in, the rule needs manual rewriting. An ML model can be retrained with new data.
4. **Rules lack uncertainty.** A rule says "go here." ML says "go here with 85% confidence; this other district with 60% confidence is a backup option."

**Why ML is better here:**

- **Captures non-linear relationships:** Demographics affect voting in complex ways. Gradient Boosting or Random Forest can learn these without me specifying them.
- **Provides feature importance:** ML can tell Chris *why* — e.g., "Internet access in Salta turned out to be more predictive than income."
- **Scales:** The same pipeline works for any province, not just Salta.
- **Supports the real action:** Chris gets a ranked list with confidence scores, allowing him to allocate canvassing teams, mailers, and candidate visits in priority order.
